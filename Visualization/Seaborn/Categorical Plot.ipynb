```python
df = sns.load_dataset('tips')
df
```

```python
sns.barplot(x=df['sex'],y=df['total_bill'])
```

```python
sns.boxplot(x='tip',y='day',data=df,palette='rainbow')
```

## Box plot vs violin plot
Box plots and violin plots are powerful, complementary visualizations for displaying the distribution of numerical data. Box plots highlight summary statistics (median, quartiles, outliers), while violin plots combine box plots with kernel density estimation to show the data's shape and density, revealing multimodal distributions that box plots might hide

```python
sns.violinplot(x='day',y='total_bill',data=df)
```

## striplot and swarmplot
Strip plots and swarm plots are categorical scatter plots used in data visualization to display the distribution of individual data points across different categories. While they are very similar, their primary difference lies in how they handle overlapping points.

```python
sns.stripplot(x='day',y='total_bill',data=df)
```

```python
sns.swarmplot(x='day',y='total_bill',data=df)
```

```python
sns.violinplot(x='day',y='tip',data=df)
sns.swarmplot(x='day',y='tip',data=df)
```
