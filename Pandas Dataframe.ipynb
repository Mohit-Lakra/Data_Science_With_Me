# Dataframes in Padas
A Pandas DataFrame is a 2 dimensional data structure, like a 2 dimensional array, or a table with rows and columns.


```python
import numpy as np
import pandas as pd
```

## Creating a Dataframe

```python
data = {
    "Name"   : ["Mohit","Rohit","Sohit","Mayank"],
    "Age"    : [19, 24, 30, 20],
    "City"   : ["Delhi", "Jaipur", "Ranchi", "Gurgaon"],
    "Salary" : [20000,22000,250000,430000]
}
df_alpha = pd.DataFrame(data)
df_alpha
```

```python
data_list = [["Mohit","Rohit","Sohit","Mayank"],[19, 24, 30, 20],["Delhi", "Jaipur", "Ranchi", "Gurgaon"],[20000,22000,250000,430000]]
pd.DataFrame(data_list)
```

```python
data_list = [
    ['Mohit', 19, 'Delhi', 20000],
    ['Rohit', 24, 'Jaipur', 22000],
    ['Sohit', 30, 'Ranchi', 250000],
    ['Mayank', 20, 'Gurgaon', 430000]
]
pd.DataFrame(data_list)
```

```python
columns=['Name','Age','City','Salary']
df = pd.DataFrame(data_list,columns=columns)
df
```

## Selection and Indexing of Columns

```python
df['Name']
```

```python
df[['Name','City']]
```

## Creating a new Column

```python
df["Designation"] = ['Doctor','Engineer','Teacher','Influencer']
df
```

## Removing Columns

```python
df.drop('Age',axis=1)
df # We can see No change
```

```python
df.drop('Age',axis=1,inplace=True)
df
```

```python
df.drop(2,inplace=True)
df
```
