# Pandas Series
A series is a one dimensional labeled array capable of holding any data type. The axis labels are collectively called index

## Creating Series
There are multiple ways to create a pandas Series

```python
import numpy as np
import pandas as pd
```

```python
labels = ['a','b','c']
l = [10,20,30]
array = np.array([10,20,30])
d = {1:10,2:20,3:30}
```

```python
pd.Series(l)
```

```python
pd.Series(l,index=labels)
```

```python
pd.Series(array)
```

```python
pd.Series(d)
```
